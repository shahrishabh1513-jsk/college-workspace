Problem Statement – 1

Aim: To implement the Vigenère Cipher algorithm for encrypting and decrypting a plaintext message using a keyword.

---



```python
# Vigenere Cipher - Encryption and Decryption

def generate_key(text, key):
    key = key.upper()
    text = text.upper()

    key_list = list(key)

    while len(key_list) < len(text):
        key_list.extend(list(key))

    return "".join(key_list[:len(text)])


def encrypt(plaintext, key):
    plaintext = plaintext.upper().replace(" ", "")
    key = generate_key(plaintext, key)

    cipher = ""

    for p, k in zip(plaintext, key):
        x = (ord(p) - ord('A') + ord(k) - ord('A')) % 26
        cipher += chr(x + ord('A'))

    return cipher


def decrypt(ciphertext, key):
    ciphertext = ciphertext.upper().replace(" ", "")
    key = generate_key(ciphertext, key)

    plain = ""

    for c, k in zip(ciphertext, key):
        x = (ord(c) - ord('A') - (ord(k) - ord('A')) + 26) % 26
        plain += chr(x + ord('A'))

    return plain


# Main Program

plaintext = input("Enter Plaintext : ")
keyword = input("Enter Keyword : ")

ciphertext = encrypt(plaintext, keyword)
print("\nEncrypted Text :", ciphertext)

decrypted = decrypt(ciphertext, keyword)
print("Decrypted Text :", decrypted)
```

: To simulate secure communication using the Vigenère Cipher with a shared secret key.

```python
# Secure Communication using Vigenere Cipher

def generate_key(text, key):
    key = key.upper()
    text = text.upper()

    key_list = list(key)

    while len(key_list) < len(text):
        key_list.extend(list(key))

    return "".join(key_list[:len(text)])


def encrypt(message, key):
    message = message.upper().replace(" ", "")
    key = generate_key(message, key)

    cipher = ""

    for m, k in zip(message, key):
        x = (ord(m) - ord('A') + ord(k) - ord('A')) % 26
        cipher += chr(x + ord('A'))

    return cipher


def decrypt(cipher, key):
    cipher = cipher.upper().replace(" ", "")
    key = generate_key(cipher, key)

    message = ""

    for c, k in zip(cipher, key):
        x = (ord(c) - ord('A') - (ord(k) - ord('A')) + 26) % 26
        message += chr(x + ord('A'))

    return message


print("----- Sender -----")

message = input("Enter Message : ")
keyword = input("Enter Shared Secret Key : ")

cipher = encrypt(message, keyword)

print("\nCiphertext Sent :", cipher)

print("\n----- Receiver -----")

received = decrypt(cipher, keyword)

print("Received Ciphertext :", cipher)
print("Decrypted Message :", received)
```

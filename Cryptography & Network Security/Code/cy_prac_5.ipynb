### TITLE: Demonstration of DES Encryption using Python

Problem Statement – 1

Aim: To implement the DES (Data Encryption Standard) algorithm for encrypting and decrypting a plaintext message using Python.

---

```python
from Crypto.Cipher import DES
from Crypto.Util.Padding import pad, unpad

# Secret key (8 bytes)
key = b'ABCDEFGH'

# Create DES cipher
cipher = DES.new(key, DES.MODE_ECB)

# Input plaintext
plaintext = input("Enter Plaintext: ")

# Convert into bytes
data = plaintext.encode()

# Apply padding
padded_data = pad(data, DES.block_size)

# Encrypt
ciphertext = cipher.encrypt(padded_data)

print("\nEncrypted Ciphertext (Bytes):")
print(ciphertext)

print("\nEncrypted Ciphertext (Hex):")
print(ciphertext.hex())

# Create DES cipher for decryption
decipher = DES.new(key, DES.MODE_ECB)

# Decrypt
decrypted = decipher.decrypt(ciphertext)

# Remove padding
original = unpad(decrypted, DES.block_size)

print("\nDecrypted Text:", original.decode())
```

Problem Statement – 2

Aim: To simulate secure communication using DES encryption with a shared secret key.


---



```python
from Crypto.Cipher import DES
from Crypto.Util.Padding import pad, unpad

# Shared Secret Key
key = b'ABCDEFGH'

print(" Sender :")

message = input("Enter Message :→")

cipher = DES.new(key, DES.MODE_ECB)

ciphertext = cipher.encrypt(
    pad(message.encode(), DES.block_size)
)

print("\nCiphertext Sent (Hex):")
print(ciphertext.hex())

print("\n Receiver :→")

decipher = DES.new(key, DES.MODE_ECB)

received = unpad(
    decipher.decrypt(ciphertext),
    DES.block_size
)

print("Received Ciphertext :")
print(ciphertext.hex())

print("Decrypted Message :")
print(received.decode())
```

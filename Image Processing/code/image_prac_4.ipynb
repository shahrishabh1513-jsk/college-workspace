```python
import cv2
from matplotlib import pyplot as plt
```

```python
# Read grayscale image
img = cv2.imread("/content/RISHABH_15.png", cv2.IMREAD_GRAYSCALE)
```

```python
# Apply a 3×3 Mean (Average) Filter on the input image
# Mean Filter
mean = cv2.blur(img, (3, 3))
```

```python
# Median Filter
median = cv2.medianBlur(img, 3)
```

```python
# Apply a 3×3 Gaussian Filter
# Gaussian Filter
gaussian = cv2.GaussianBlur(img, (3, 3), 0)
```

```python
# Laplacian Filter (with larger kernel size)
laplacian = cv2.Laplacian(img, cv2.CV_64F, ksize=5)
laplacian_abs = cv2.convertScaleAbs(laplacian)
```

```python
# Apply threshold to make edges more visible
# The threshold value (50) may be adjusted depending on the image
_, laplacian_thresholded = cv2.threshold(
    laplacian_abs,
    50,
    255,
    cv2.THRESH_BINARY
)
```

```python
# Display Results
titles = [
    "Original",
    "Mean Filter",
    "Median Filter",
    "Gaussian Filter",
    "Laplacian Filter",
    "Laplacian Filter "
]
```

```python
images = [
    img,
    mean,
    median,
    gaussian,
    laplacian_abs,
    laplacian_thresholded
]
```

```python
plt.figure(figsize=(15, 10))
```

```python
for i in range(len(images)):
    plt.subplot(2, 3, i + 1)
    plt.imshow(images[i], cmap="gray")
    plt.title(titles[i])
    plt.axis("off")
```

```python
plt.tight_layout()
plt.show()
```
